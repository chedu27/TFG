```python
#1. Data overview
import pandas as pd

# Para ver todas las columnas
pd.set_option('display.max_columns', None)

# Para ver todas las filas
#pd.set_option('display.max_rows', None)

#1.1 Cargar el dataset
df = pd.read_excel("Telco_customer_churn.xlsx")

#1.2 Dimensión del dataset
df.shape
```

```python
#1.3 Ver las primeras filas
df.head()
```

```python
#1.4 Información general del dataset
df.info()
```

```python
#convertir Total Charges a string
# 1. Convertir a float (errors='coerce' transforma lo que no es número en NaN)
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce')

#Muestra las filas con valor nulo
# Filtrar y mostrar las filas donde Total Charges es NaN
filas_nan = df[df["Total Charges"].isna()]
display(filas_nan)

#contarlas:
print(f"Total de NaNs: {df['Total Charges'].isna().sum()}")
```

```python
# 2. Rellenar los valores nulos con 0
df['Total Charges'] = df['Total Charges'].fillna(0)
# Muestra las filas donde Total Charges es 0
df_ceros = df[df["Total Charges"] == 0]
display(df_ceros)
```

```python
#5.1 Identificar variables categóricas
cat_cols = df.select_dtypes(include=["str"]).columns
cat_cols
```

```python
#5.2 Análisis básico
for col in cat_cols:
    print(df[col].value_counts())
    print("\n")
```

```python
# Separar variables binarias vs multiclase
# Variables binarias:
binary_cols = [col for col in cat_cols if df[col].nunique() == 2]
#5.3 Churn rate por variable
for col in binary_cols:
    print(df.groupby(col)["Churn Value"].mean().sort_values(ascending=False))
    print('\n')
```

```python
#Variables multicase:
multi_cols = [col for col in cat_cols if df[col].nunique() > 2]

#5.3 Churn rate por variable
for col in multi_cols:
    print(df.groupby(col)["Churn Value"].mean().sort_values(ascending=False))
    print('\n')
```

```python
#6.1 Combinación Contract + Internet Service + Tech Support
df.groupby([
    "Contract",
    "Internet Service",
    "Tech Support",
])["Churn Value"].mean().sort_values(ascending=False).head(10)
```

```python
#6.4 Contract + Tech Support + Online Security
df.groupby([
    "Contract",
    "Tech Support",
    "Online Security",
])["Churn Value"].mean().sort_values(ascending=False).head(10)
```
